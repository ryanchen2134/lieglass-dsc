```python
import glob, subprocess
total_seconds = 0.0
durations = []
# for path in sorted(glob.glob('*.mp4')): do all recursive subdirs, like **/*.mp4
for path in sorted(glob.glob('**/*.mp4', recursive=True)):
    try:
        print('checking', path)
        cmd = ['ffprobe','-v','error','-show_entries','format=duration',
               '-of','default=noprint_wrappers=1:nokey=1', path]
        out = subprocess.check_output(cmd).decode().strip()
        d = float(out)
    except Exception:
        try:
            from moviepy.editor import VideoFileClip
            d = VideoFileClip(path).duration
        except Exception as e:
            print('skipping', path, '(', e, ')')
            continue
    durations.append(d)
    total_seconds += d
print(f'Found {len(durations)} .mp4 files; total seconds = {total_seconds:.2f}')
```
